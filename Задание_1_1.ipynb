### 1. Объекты-переменные, целые числа, системы счисления и преобразования типов
Переменные в Python являются именами, которые ссылаются на значения в памяти компьютера. Для присваивания значения используется символ `=`. Python имеет встроенную поддержку целых чисел и чисел с плавающей точкой. Также язык позволяет выражать числа в двоичной, восьмеричной и шестнадцатеричной системах счисления и менять типы данных с помощью таких встроенных функций, как `int()`.

```python
# Объекты-переменные и тип
a = 7
print("Значение переменной a:", a)
b = a
print("Значение переменной b:", b)
print("Тип переменной a:", type(a))

# Арифметические операции с целыми числами
print("Сложение (5+9):", 5 + 9)
print("Деление с плавающей точкой (9/5):", 9 / 5)
print("Целочисленное деление (9//5):", 9 // 5)
print("Остаток от деления (13%5):", 13 % 5)

# Системы счисления
print("Двоичная система (0b10):", 0b10)
print("Восьмеричная система (0o10):", 0o10)
print("Шестнадцатеричная система (0xf):", 0xf)

# Преобразование типов
print("Преобразование bool в int (True):", int(True))
print("Преобразование float в int (98.6):", int(98.6))
print("Преобразование строки в int ('99'):", int('99'))
```

### 2. Списки и кортежи
Списки служат для хранения объектов в определенном порядке, особенно если их содержимое может изменяться. Кортежи, как и списки, являются последовательностями произвольных элементов, но в отличие от списков они неизменяемы (вы не можете добавить, удалить или изменить элементы после создания).

```python
# Списки
weekdays = ['Понедельник', 'Вторник', 'Среда', 'Четверг']
print("Исходный список:", weekdays)

# Получение и изменение элементов
print("Первый элемент списка:", weekdays[0])
weekdays[0] = 'Воскресенье'
print("Измененный список:", weekdays)

# Добавление элементов
weekdays.append('Пятница')
print("Список после добавления (append):", weekdays)

# Извлечение среза
print("Срез списка [1:3]:", weekdays[1:3])

# Кортежи
physicists_tuple = ('Шелдон', 'Леонард', 'Радж')
print("Кортеж:", physicists_tuple)

# Распаковка кортежа
a, b, c = physicists_tuple
print("Распакованные значения:", a, b, c)
```

### 3. Массивы NumPy и математика массивов
Модуль `NumPy` используется для выполнения научных расчетов и предоставляет доступ к быстрым многомерным массивам (ndarray). В отличие от списков Python, все элементы массива NumPy должны иметь одинаковый тип. Модуль позволяет легко выполнять математические операции над всеми элементами массива одновременно без использования циклов.

```python
import numpy as np

# Создание массивов
arr = np.array([1, 2, 3, 4])
print("Одномерный массив NumPy:", arr)

# Использование zeros
zeros_arr = np.zeros(3)
print("Массив из нулей:", zeros_arr)

# Изменение формы массива (reshape) и генерация (arange)
arr_reshaped = np.arange(6).reshape(2, 3)
print("Массив формы 2x3:\n", arr_reshaped)

# Математика массивов
arr2 = np.array([5, 6, 7, 8])
print("Сложение массивов:", arr + arr2)
print("Умножение элементов массива на число (arr * 10):", arr * 10)
```

### 4. Словари
Словарь очень похож на список, но порядок элементов в нем не имеет значения. Вместо числовых индексов (смещений) для каждого значения указывается уникальный ключ (например, строка). Словари можно изменять: добавлять, удалять и перезаписывать элементы "ключ-значение".

```python
# Создание словаря
gryffindor = {
    'Поттер': 'Гарри',
    'Дамблдор': 'Альбус',
    'Грейнджер': 'Гермиона'
}
print("Исходный словарь:", gryffindor)

# Добавление и изменение элементов
gryffindor['Уизли'] = 'Рон'
print("Словарь после добавления:", gryffindor)

# Получение элементов по ключу и с помощью get()
print("Значение по ключу 'Грейнджер':", gryffindor['Грейнджер'])
print("Использование get() для несуществующего ключа:", gryffindor.get('Малфой', 'Не учился на факультете'))

# Получение ключей и значений
print("Все ключи:", list(gryffindor.keys()))
print("Все значения:", list(gryffindor.values()))
```

### 5. Создание крупных структур данных
В Python можно объединять базовые встроенные структуры данных (списки, кортежи, словари) в собственные более крупные и сложные структуры. Например, можно создать список, состоящий из списков, или словарь, содержащий в качестве значений списки.

```python
# Исходные списки
friends = ['Чендлер', 'Росс', 'Джо']
physicists = ['Леонард', 'Шелдон', 'Радж']

# Создание списка списков
list_of_lists = [friends, physicists]
print("Список списков:\n", list_of_lists)
print("Доступ к элементу списка списков (list_of_lists[0][1]):", list_of_lists[0][1])

# Создание словаря из списков
dict_of_lists = {
    'Друзья': friends,
    'Физики': physicists
}
print("\nСловарь из списков:\n", dict_of_lists)
print("Доступ к списку физиков через ключ словаря:", dict_of_lists['Физики'])
```
